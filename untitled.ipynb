```python
import pandas as pd
df = pd.read_csv('./results/results_flash-holmes.csv')
df_full = df[df['encoding'] == 'full']
df_half = df[df['encoding'] == 'half']
grouped_avg_full = df_full.groupby(["probing_dataset", "model_name"])["score"].mean().reset_index()
grouped_avg_half = df_half.groupby(["probing_dataset", "model_name"])["score"].mean().reset_index()
grouped_avg_full.to_csv('./results/averaged_flash-holmes.csv')
grouped_avg_half.to_csv('./results/averaged_flash-holmes_halfp.csv')
```
