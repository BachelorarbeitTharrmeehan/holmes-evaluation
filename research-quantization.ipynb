```python
import pandas as pd
import plotly.express as px

def generate_custom_plots(dataframe, column_name='linguistic subfield', n=15):
    """
    Generates custom plots for top and bottom n models grouped by a specified column.
    
    Parameters:
    - dataframe: pd.DataFrame - The input dataframe containing model data.
    - column_name: str - The column to group by (replaces 'linguistic subfield').
    - n: int - The number of top and bottom entries to filter for each group.
    
    Returns:
    - None: Displays the generated plots.
    """
    def filter_top_bottom(df, top=True, n=n):
        filtered_dfs = []
        for group in df[column_name].unique():
            group_df = df[df[column_name] == group]
            if top:
                filtered_dfs.append(group_df.nlargest(n, 'score'))
            else:
                filtered_dfs.append(group_df.nsmallest(n, 'score'))
        return pd.concat(filtered_dfs)

    # Filter top and bottom models
    top_models = filter_top_bottom(dataframe, top=True, n=n)
    bottom_models = filter_top_bottom(dataframe, top=False, n=n)

    # Create steps for arrow key control
    def create_steps(df):
        return [
            dict(
                label=group,
                method="animate",
                args=[
                    [group],
                    dict(
                        mode="immediate",
                        frame={"duration": 500, "redraw": True},
                        transition={"duration": 300},
                    ),
                ],
            )
            for group in df[column_name].unique()
        ]

    # Plot for Top Models
    fig_top = px.bar(
        top_models,
        x='score',
        y='model_name',
        color='score',
        animation_frame=column_name,
        title=f'Top {n} Models for Each {column_name.capitalize()}',
        labels={'model_name': 'Model Name', 'score': 'Score'},
        height=1000,
        width=1300
    )

    fig_top.update_layout(
        xaxis_title="Score",
        yaxis_title="Model Name",
        title_x=0.5,
        margin=dict(l=150, r=40, t=80, b=80),
        yaxis=dict(
            categoryorder='total ascending'  # Sort highest score at the top
        ),
        sliders=[
            dict(
                active=0,
                pad={"t": 50},
                len=0.9,
                currentvalue={"font": {"size": 20}},
                steps=create_steps(top_models)  # Add steps for arrow key navigation
            )
        ]
    )

    # Plot for Bottom Models
    fig_bottom = px.bar(
        bottom_models,
        x='score',
        y='model_name',
        color='score',
        animation_frame=column_name,
        title=f'Bottom {n} Models for Each {column_name.capitalize()}',
        labels={'model_name': 'Model Name', 'score': 'Score'},
        height=1000,
        width=1300
    )

    fig_bottom.update_layout(
        xaxis_title="Score",
        yaxis_title="Model Name",
        title_x=0.5,
        margin=dict(l=150, r=40, t=80, b=80),
        yaxis=dict(
            categoryorder='total descending'  # Sort lowest score at the top
        ),
        sliders=[
            dict(
                active=0,
                pad={"t": 50},
                len=0.9,
                currentvalue={"font": {"size": 20}},
                steps=create_steps(bottom_models)  # Add steps for arrow key navigation
            )
        ]
    )

    # Ensure unnecessary buttons are removed
    fig_top["layout"].pop("updatemenus", None)
    fig_bottom["layout"].pop("updatemenus", None)

    # Show both plots
    fig_top.show()
    fig_bottom.show()
```

# Statistical Calculation
Null Hypothesis = There is 0 difference between quantized and non-quantized model 

Alternative Hypothesis = There is some difference between quantized and non-quantized model, to be exact the quantized model has got a worse performance than the unquantized one

p-value < 0.05 => Rejection of our Null Hypothesis = Statistical Significance

## Only Sentence and Sentence Pair Classification (Probe Type)

```python
# Specify the two models you want to compare
model_1 = "Qwen/Qwen2.5-0.5B-Instruct-AWQ"
model_2 = "Qwen/Qwen2.5-0.5B-Instruct"

# Filter the data for the two models
model_1_scores = final_set_sorted[final_set_sorted['model_name'] == model_1]['score']
model_2_scores = final_set_sorted[final_set_sorted['model_name'] == model_2]['score']

# Perform the t-test
t_stat, p_value = ttest_ind(model_1_scores, model_2_scores)

# Display the results
print(f"T-statistic: {t_stat}")
print(f"P-value: {p_value}")

# Interpret the result
if p_value < 0.05:
    print("There is a statistically significan t difference between the two models.")
else:
    print("There is no statistically significant difference between the two models.")
```

```python
from scipy.stats import levene

# Perform Levene's test
stat, p_value = levene(model_1_scores, model_2_scores)

# Display the results
print(f"Levene's Test Statistic: {stat}")
print(f"P-value: {p_value}")

if p_value < 0.05:
    print("Variances are significantly different (heteroscedasticity).")
else:
    print("No significant difference in variances (homoscedasticity).")
```

```python
from scipy.stats import bartlett

# Perform Bartlett's test
stat, p_value = bartlett(model_1_scores, model_2_scores)

# Display the results
print(f"Bartlett's Test Statistic: {stat}")
print(f"P-value: {p_value}")

if p_value < 0.05:
    print("Variances are significantly different (heteroscedasticity).")
else:
    print("No significant difference in variances (homoscedasticity).")
```

# Statiscial T-Tests between all model combinations

```python
from itertools import combinations
from scipy.stats import ttest_ind
import pandas as pd


models = final_set_sorted['model_name'].unique()

results = []

for model_1, model_2 in combinations(models, 2):
    model_1_scores = final_set_sorted[final_set_sorted['model_name'] == model_1]['score']
    model_2_scores = final_set_sorted[final_set_sorted['model_name'] == model_2]['score']
    
    t_stat, p_value = ttest_ind(model_1_scores, model_2_scores)
    
    results.append({
        'Model 1': model_1,
        'Model 2': model_2,
        'T-Statistic': t_stat,
        'P-Value': p_value,
        'Significant': p_value < 0.05
    })

results_df = pd.DataFrame(results)

results_df = results_df.sort_values(by='P-Value')
```

```python
def calculate_score_change(final_set, model_name_x, model_name_y, column_group='linguistic phenomena'):
    one = final_set[final_set['model_name'] == model_name_x]
    two = final_set[final_set['model_name'] == model_name_y]

    three = pd.merge(one, two, on=column_group)

    three['score_change_pct'] = (three['score_y'] - three['score_x']) / three['score_x'] * 100

    three = three.fillna(0).sort_values(by='score_change_pct')

    return three
```
